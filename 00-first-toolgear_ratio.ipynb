```python
teeth_in = int(input("Teeth on input gear: "))
teeth_out = int(input("Teeth on output gear: "))
rpm_in = float(input("Input speed (rpm): "))
torque_in=float(input("Input torque (N.m): "))

ratio = teeth_out / teeth_in
rpm_out = rpm_in / ratio
torque_out=torque_in*ratio

print(f"Gear ratio: {ratio:.2f}")
print(f"Output speed: {rpm_out:.1f} rpm")
print(f"Output torque: {torque_out:.1f} N.m")
```
